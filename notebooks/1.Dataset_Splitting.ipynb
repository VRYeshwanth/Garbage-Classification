# Dataset Splitting

```python
import os
import shutil
import random

random.seed(42)

ORIGINAL_DATASET = '../standardized_384'
classes = os.listdir(ORIGINAL_DATASET)
print(f"Present Classes: {classes}")

DEST_DATASET = '../dataset'
os.makedirs(DEST_DATASET)

print("Starting dataset splitting...")

for class_name in classes:
    directory = os.path.join(ORIGINAL_DATASET, class_name)
    files = os.listdir(directory)
    random.shuffle(files)

    train_dir = os.path.join(DEST_DATASET, 'train', class_name)
    val_dir = os.path.join(DEST_DATASET, 'val', class_name)
    test_dir = os.path.join(DEST_DATASET, 'test', class_name)

    os.makedirs(train_dir)
    os.makedirs(val_dir)
    os.makedirs(test_dir)

    n = len(files)
    train_end = int(n * 0.7)
    val_end = train_end + int(n * 0.2)

    train_files = files[:train_end]
    val_files = files[train_end:val_end]
    test_files = files[val_end:]

    for train_file in train_files:
        shutil.copy(os.path.join(directory, train_file), train_dir)
    for val_file in val_files:
        shutil.copy(os.path.join(directory, val_file), val_dir)
    for test_file in test_files:
        shutil.copy(os.path.join(directory, test_file), test_dir)

    print(f"\nClass name: {class_name}")
    print(f"No. of images in train: {len(os.listdir(train_dir))}")
    print(f"No. of images in val: {len(os.listdir(val_dir))}")
    print(f"No. of images in test: {len(os.listdir(test_dir))}")
```
